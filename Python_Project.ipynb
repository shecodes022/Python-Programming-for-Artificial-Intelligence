**Q4: Complete the code for following three functions with appropriate docstring, as described below.**

* ```sum_with_none```: function that takes input x as a list or numpy array and returns sum of all the numbers in it as sum_x, excluding None values or any other kinds of None (np.nan). It also returns number of values that are added n_x.
  * For example, if x = [1,2,3, None, 3, np.nan] is passed then function should return sum_x=9, and n_x =4.


* ```mean_with_none```: function that takes input x as a list or numpy array and returns mean of all the numbers in it, excluding None values or any other kinds of None (np.nan).  Try re-using 'sum_with_none' function.


* ```std_with_none```: function that takes input as a list or numpy array and returns standard deviation of all the numbers in it, excluding None values or any other kinds. Try re-using 'sum_with_none' and/or  'mean_with_none' functions.


**For each function, Do add a doctring, explaining your code logic, input and output. This doctring is called in 'YOUR EXPLAINATION' section to check how well your function is documented.**

````python
def sum_with_none(x):
    '''
    This function calculates the sum of values in a numpy array or list,
    ignoring missing values such as None and np.nan.

     Parameters:
    x : list or numpy array
        Input data containing numbers and possibly missing values.

    Returns:
    sum_x : float
        Sum of valid numeric values.
    n_x : int
        Number of valid values used in the sum.

    Logic:
    - Loop through each value in x
    - Skip values that are None or np.nan
    - Add valid values to sum
    - Count number of valid values

    '''
    sum_x = 0
    n_x = 0

    for value in x:
        if value is None:
            continue

        try:
            if np.isnan(value):
                continue
        except:
            pass

        sum_x += value
        n_x += 1

    return sum_x, n_x


def mean_with_none(x):
    '''
     This function calculates the mean (average) of values in a numpy array or list,
    ignoring missing values such as None and np.nan.

    Parameters:
    x : list or numpy array
        Input data containing numbers and possibly missing values.

    Returns:
    mean_x : float
        Mean of valid values.
        Returns np.nan if no valid values exist.

    Logic:
    - Use sum_with_none to get total sum and count
    - Divide sum by count
    - If count is zero, return np.nan
    '''
    mean_x = None

    sum_x, n_x = sum_with_none(x)

    if n_x == 0:
        mean_x = np.nan
    else:
        mean_x = sum_x / n_x

    return mean_x


def std_with_none(x):
    '''
    This function calculates the standard deviation of values in a numpy array or list,
    ignoring missing values such as None and np.nan.

    Parameters:
    x : list or numpy array
        Input data containing numbers and possibly missing values.

    Returns:
    std_x : float
        Standard deviation of valid values.
        Returns np.nan if no valid values exist.

    Logic:
    - First calculate the mean using mean_with_none
    - Loop through values and compute squared difference from mean
    - Ignore None and np.nan values
    - Compute variance (average squared difference)
    - Take square root to get standard deviation

    '''
    std_x = None

    mean_x = mean_with_none(x)

    if isinstance(mean_x, float) and np.isnan(mean_x):
        return np.nan

    sum_sq_diff = 0
    n_x = 0

    for value in x:
        if value is None:
            continue

        try:
            if np.isnan(value):
                continue
        except:
            pass

        sum_sq_diff += (value - mean_x) ** 2
        n_x += 1

    if n_x == 0:
        return np.nan

    variance = sum_sq_diff / n_x
    std_x = np.sqrt(variance)

    return std_x
````

````python
#Q6A

education_map = {
    'No School': 0,
    'Some School': 1,
    'High School': 2,
    'Undergraduate Degree (UG)': 3,
    "Postgraduate / Master’s Degree (PG)": 4,
    "Postgraduate / Master's Degree (PG)": 4,
    'Doctorate Degree (PhD)': 5
}

def bool_to_num(x):
    """
    Convert boolean values into numbers for distance calculation.

    Parameters:
    x : bool or any
        Input value from LikeVollyball or LikeTableTennis.

    Returns:
    float
        Returns 1.0 for True, 0.0 for False, and np.nan otherwise.
    """
    if x is True:
        return 1.0
    elif x is False:
        return 0.0
    else:
        return np.nan


def row_to_numeric(row):
    """
    Convert one row from the dataset into a numeric numpy array
    using the selected attributes for closeness computation.

    Parameters:
    row : numpy array
        One row from data_np.

    Returns:
    numpy array
        Numeric feature vector containing:
        Age, Height, Weight, Education score, NumberOfLanguages,
        LikeVollyball, LikeTableTennis, HappinessLevel.

    Logic:
    - Extract only numeric or converted-to-numeric attributes
    - Convert education into ordinal numeric score
    - Convert booleans into 0/1
    - Keep missing values as np.nan
    """

    age = row[4]
    height = row[5]
    weight = row[6]
    education = row[8]
    languages = row[9]
    like_vollyball = row[10]
    like_table_tennis = row[11]
    happiness = row[13]

    edu_score = education_map.get(education, np.nan)
    volly_score = bool_to_num(like_vollyball)
    table_score = bool_to_num(like_table_tennis)

    return np.array([
        age,
        height,
        weight,
        edu_score,
        languages,
        volly_score,
        table_score,
        happiness
    ], dtype=float)


def get_feature_min_max(data_np):
    """
    Compute column-wise minimum and maximum values for the selected
    numeric attributes across the whole dataset.

    Parameters:
    clean_np : numpy array
        Cleaned dataset.

    Returns:
    mins : numpy array
        Minimum value of each selected attribute.
    maxs : numpy array
        Maximum value of each selected attribute.

    Logic:
    - Convert each row into numeric feature vector
    - Stack all rows together
    - Use np.nanmin and np.nanmax to ignore missing values
    """

    all_rows = []

    for i in range(data_np.shape[0]):
        numeric_row = row_to_numeric(data_np[i])
        all_rows.append(numeric_row)

    all_rows = np.array(all_rows, dtype=float)

    mins = np.nanmin(all_rows, axis=0)
    maxs = np.nanmax(all_rows, axis=0)

    return mins, maxs


def closeness_between_two_people(row1, row2, mins, maxs):
    """
    Compute the closeness between two people using normalized Euclidean distance.

    Parameters:
    row1 : numpy array
        First row from clean_np.
    row2 : numpy array
        Second row from clean_np.
    mins : numpy array
        Minimum values of selected attributes in dataset.
    maxs : numpy array
        Maximum values of selected attributes in dataset.

    Returns:
    distance : float
        Normalized Euclidean distance between the two people.
        Smaller distance means more similar / closer.
        Returns np.nan if no valid common attributes exist.

    Logic:
    - Convert both rows into numeric vectors
    - Apply Min-Max normalization to scale values between 0 and 1
    - Ignore attributes where either value is missing
    - Compute Euclidean distance only on common valid attributes
    - Divide by number of valid dimensions inside sqrt so comparison
      stays fair when some values are missing
    """

    x1 = row_to_numeric(row1)
    x2 = row_to_numeric(row2)

    x1_scaled = (x1 - mins) / (maxs - mins)
    x2_scaled = (x2 - mins) / (maxs - mins)

    valid_mask = ~np.isnan(x1_scaled) & ~np.isnan(x2_scaled)

    n_valid = np.sum(valid_mask)

    if n_valid == 0:
        return np.nan

    diff = x1_scaled[valid_mask] - x2_scaled[valid_mask]

    distance = np.sqrt(np.sum(diff ** 2) / n_valid)

    return distance
````

````python
#Q6B
mins, maxs = get_feature_min_max(data_np)

best_distance = np.inf
best_i = None
best_j = None

for i in range(data_np.shape[0]):
    for j in range(i + 1, data_np.shape[0]):
        d = closeness_between_two_people(data_np[i], data_np[j], mins, maxs)

        if not np.isnan(d) and d < best_distance:
            best_distance = d
            best_i = i
            best_j = j

print("Closest distance:", best_distance)
print("Row index 1:", best_i)
print("Row index 2:", best_j)

print("\nFirst person:")
print(data_np[best_i])

print("\nSecond person:")
print(data_np[best_j])
````
